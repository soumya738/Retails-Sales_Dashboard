```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
df = pd.read_csv("store_sales_data (2).csv")
print("Original shape:", df.shape)
df.head()
```

```python
date_cols = ["Date of Birth", "Sales Date", "Order Date", "Ship Date"]
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors="coerce")
df.head()
```

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Order Month Name"] = df["Order Date"].dt.strftime("%B")
df["Order Quarter"] = df["Order Date"].dt.quarter
df["Order Quarter Name"] = "Q" + df["Order Quarter"].astype(str)
df["Order Weekday"] = df["Order Date"].dt.day_name()
df["Order YearMonth"] = df["Order Date"].dt.to_period("M").astype(str)
```

```python
df["Customer Full Name"] = df["Customer Name"] + " " + df["Last Name"]
```

```python
df["Age"] = 2026 - df["Date of Birth"].dt.year
```

```python
df["Profit Margin %"] = (df["Profit"] / df["Sales"]).round(4)
```

```python
df["Shipping Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
df.columns = [c.strip().replace(" ", "_") for c in df.columns]
```

```python
final_cols = [
    "Customer_ID", "Customer_Full_Name", "Age", "Segment",
    "Order_ID", "Order_Date", "Order_Year", "Order_Month", "Order_Month_Name",
    "Order_Quarter", "Order_Quarter_Name", "Order_Weekday", "Order_YearMonth",
    "Ship_Date", "Ship_Mode", "Shipping_Days",
    "Sales", "Quantity", "Discount", "Profit", "Profit_Margin_%",
    "Product_ID", "Product_Name", "Sub-Category", "Category_of_Goods",
    "Outlet_Type", "City_Type", "Region", "State", "Postal_Code"
]
```

```python
df_clean = df[final_cols].copy()
```

```python
df_clean.to_csv("store_sales_cleaned.csv", index=False)
df_clean.to_parquet("store_sales_cleaned.parquet", index=False)  # faster for Power BI

print("Cleaned shape:", df_clean.shape)
print("Saved: store_sales_cleaned.csv and store_sales_cleaned.parquet")
print(df_clean.head(3))
```
